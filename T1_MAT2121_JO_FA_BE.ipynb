## Pregunta 2

2. En un mismo grafico, trace las tres funciones considerando que se pronostican 12 horas de lluvia.

## Funciones

2.1. **Valparaíso**: $$V(t)=0,8t^{2}+12,8t+37$$


2.2. **Santiago**: $$S(t) =t^{2}+16,5t+30$$


2.3. **Concepción**: $$C(t) = 1,2t^{2}+19,2t+25$$




```python
import numpy as np
import matplotlib.pyplot as plt

def V(t):
    return 0.8*t**2 + 12.8*t + 37

def S(t):
    return t**2 + 16.5*t + 30

def C(t):
    return 1.2*t**2 + 19.2*t + 25

t = np.arange(0, 12, 0.01)
plt.plot(t, V(t), color = 'red', label = 'V(t)')
plt.plot(t, S(t), color = 'blue', label = 'S(t)')
plt.plot(t, C(t), color = 'green', label = 'C(t)')
plt.title('Cantidad de milimetros caidos en 12 horas')
plt.ylabel('Milimetros')
plt.xlabel('Tiempo en horas')
plt.legend()
plt.grid(True)
plt.show()

```
