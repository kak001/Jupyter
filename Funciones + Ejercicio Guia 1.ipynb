```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0, 6, 0.01)

y = 120*x+2000

plt.plot(x, y)

plt.axis([0, 6, 0, 2800])

plt.xlabel("Energia utilizada ({kWh}).")

plt.ylabel("Costo para el usuario ($).")

plt.title("Costo para el usuario segun uso de energia.")

plt.grid(True)

plt.show
```

```python
#Ejercicio 4

import matplotlib.pyplot as plt
import numpy as np

# 1.
def l(h):
    return 1.85*h

# 2.
#Variable dependiente: La longitud del hable
#Variable independiente: El tiempo transcurrido

# 3.
#El dominio de la funcion son todos los valores que toma h, en este caso las horas que va desde [0, 3567]

# 4.
h = np.arange(0, 3567, 1)

plt.plot(h, l(h))

plt.title("Relacion entre tiempo y distancia para instalar el cable submarino")
plt.xlabel("Tiempo transcurrido (h)")
plt.ylabel("Distancia recorrida (km)")

plt.grid(True)
plt.show()

# 5.
time = l(148)
meters = time*1000
print(f"Al transcurrir 148 horas, se instalan {meters} metros de cable")

time2 = l(2300)
meters2 = time2*1000
print(f"Al transcurrir 2300 horas, se instalan {meters2} metros de cable")

# 6.
h2 = 3480 / 1.85
print(f"Al instalarse 3480 kilometros de cable, transcurren {h2} horas")

# 7.
h = 6600 / 1.85
print(f"El tiempo que demoro el barco en instalar el cable son {h} horas.")
```

```python
#Ejercicio5

import matplotlib.pyplot as plt
import numpy as np


def f(t):
    return 0.4*t

def g(t):
    return 0.3*t

# 1.
#Grafico de tiempo y distancia entre metro de santiago y buses red 

t = np.arange(0, 100, 1)
plt.plot(t, f(t), label = 'f(t)')
plt.plot(t, g(t), label = 'g(t)')
plt.title('Relacion tiempo y distancia')
plt.ylabel('Kilometros recorridos (km)')
plt.xlabel('Minutos recorridos (m)')
plt.legend()
plt.grid(True)
plt.show()

# 2.
t = 1.2 * 9 + 0.5 * 8
print(f"El extremo superior del dominio es: {t:.1f}")

# 3.
#El medio de transporte mas rapido es el metro de santiago, ya que a los mismos minutos transcurridos, el metro recorre mas distancia que los buses Red, siendo mas rapido

# 4.
time = 6 / 0.4
time2 = 6 / 0.3

print(f"Tiempo demorado en metro: {time:.1f}")
print(f"Tiempo demorado en bus Red: {time2:.1f}")
```

```python
#Ejercicio 6

import matplotlib.pyplot as plt
import numpy as np

def T(t):
    return -0.5*t**2 + 3*t + 20

# 1.
#Variable dependiente: La temperatura del servidor en °C -> T(t)
#Variable independiente: El tiempo que transcurre en horas -> t

# 2.
#El dominio de la funcion T(t) va es [0, 9]

# 3.
t = np.arange(0, 9, 0.01)
plt.plot(t, T(t), color = 'red', label = "Temperatura del servidor")
plt.title('Relacion tiempo y grados')
plt.ylabel('Temperatura (°C)')
plt.xlabel('Tiempo (h)')
plt.legend()
plt.grid(True)
plt.show()

# 4.
#A las 3 horas alcanza su maxima temperatura
print(f"La temperatura que alcanza el servido a las 3 horas es: {T(3):.1f}")

# 5.
print(f"A las 13:00 horas el servidor alcanza {T(5):.1f} °C")
print(f"A las 17:00 horas, el servido alcanza {T(9):.1f} °C")
```
