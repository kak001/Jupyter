```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0, 6, 0.01)

y = 120*x+2000

plt.plot(x, y)

plt.axis([0, 6, 0, 2800])

plt.xlabel("Energia utilizada ({kWh}).")

plt.ylabel("Costo para el usuario ($).")

plt.title("Costo para el usuario segun uso de energia.")

plt.grid(True)

plt.show
```

```python
def m(s):
    return 100*s

segundo = 45
print(f"A los {segundo} segundos, se obtienen {m(segundo)} megabits.")

segundo2 = 90
print(f"A los {segundo2} segundos, se obtienen {m(segundo2)} megabits.")

segundo3 = 1*60*60
print(f"A los {segundo3} segudos cuales equivalen a una hora, se obtienen {m(segundo3)} megabits.")
```

```python
def tabla_datos():
    for i in range(0, 1100, 100):
        print(f"{i} segundos | {m(i)} megabits")

tabla_datos()
```
