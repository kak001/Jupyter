## Problema 1

Sucesion $a_{n} = 3n^{2} + 7$

1. Los primeros $5$ términos.

```python
#Lista
b = []

#Primero 5 numeros
print("Primero 5 numeros de la sucesion.")
for i in range(5):
    n = i + 1
    b.append(3*n**2 + 7)
    print(f"Posicion: {n} | Numero: {b[i]}")
```

2. Los $5$ términos que vienen inmediatamente después del decimoquinto término.

```python
#Lista 
b = []

#Los 5 terminos sucesivos despues del decimo quinto numero
print(f"5 numeros sucesivos despues del decimo quinto numero.")
for i in range(6):
    n = i + 15
    b.append(3*n**2 + 7)
    print(f"Posicion: {n} | Numero: {b[i]}")
```

## Problema 2

Sucesion $g_{n} = 5n^{3}$.

1. Los primeros $4$ términos de la sucesión.

```python
#Lista 
b = []

#Ciclo for
print("Primeros 4 terminos sucesivos.")
for i in range(4):
    n = i + 1
    b.append(5*n**3)
    print(f"Posicion: {n} | Numero: {b[i]}")
```

2. Los $4$ términos que vienen inmediatamente después del octavo término.

```python
#Lista 
b = []

#Ciclo for
for i in range(5):
    n = i + 8
    b.append(5*n**3)
    print(f"Posicion: {n} | Numero: {b[i]}")
```

3. Determinar si el término $40.000$ pertenece a la sucesión. Si es así, indicar en qué posición se encuentra.

```python
valor = 40000

n = 1

g_n = 5*n**3

while g_n < valor:
    n += 1
    g_n = 5*n**3

if g_n == valor:
    print(f"{valor} es un termino de la sucesion y se encuentra en la posicion {n}.")
else:
    print(f"{valor} no pertenece a la sucesion.")
```

## Problema 3

Sucesion 

\begin{split}
F(n) &= F(n - 1) + F(n - 2), \ \text{ para $n \geq 2$}
\end{split}

1. Implementa un código en *Python* que muestre en pantalla los primeros $20$ términos de la sucesión.

```python
#Lista
b = [0, 1]

#Primero 20 numero sucesivos de Fibonacci
for i in range(18):
    b.append(b[-1] + b[-2])

for i in range(20):
    print(f"Posicion: {i + 1} | Numero: {b[i]}")
```

2. Calcula, utilizando un ciclo *for*, la suma de los primeros $30$ términos de la sucesión de Fibonacci.

```python
b = [0, 1]

for i in range(28):
    b.append(b[-1] + b[-2])

n = 0

for i in range(30):
    n += b[i]
    print(f"Posicion: {i + 1} | Numero: {n}")
```
