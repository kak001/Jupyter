## Pregunta 2

Encuentra la función que relaciona el tamaño de la entrada con el tiempo de ejecución del algoritmo, para esto utiliza la función *polyfit* de *Numpy*.


```python
import numpy as np

unidades = np.array([100, 200, 500, 1000, 2000])
milesimas = np.array([2, 4, 10, 20, 40])

pendiente, intercepto = np.polyfit(unidades, milesimas, 1)


print(pendiente, intercepto)

print(f"La pendiente es: {pendiente:.2f}")
print(f"El intercepto es: {intercepto:.2f}")

print(f"La funcion es: ms(u) = 0.02*u")
```

## Pregunta 4

Utiliza la función *fsolve* de *Scipy* para determinar el tamaño de entrada que haría que el tiempo de ejecución sea de $50$ milisegundos.


```python
from scipy.optimize import fsolve

def U(ms):
    return 0.02*ms - 50

ms = 0

solucion2 = fsolve(U, ms)

print(solucion2)

print(f"Cuando transcurren 50 milisegundos, el tamaño de entrada sera de: {solucion2} unidades.")
```

## Ejercicio 3

3. Calcula el tiempo de carga y descarga para un archivo de $750$ MB.
